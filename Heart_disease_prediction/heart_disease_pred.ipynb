```python
data.drop('fbs',axis=1,inplace=True)
```

```python
plt.plot('target','sex',data=data,marker='o')
plt.show()
```

```python
y=data['target']
X=data.drop('target',axis=1)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_scaled=sc.fit_transform(X)
```

```python
X_scaled
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,train_size=0.7,random_state=42)
```

```python
X_train
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Activation
```

```python
model=Sequential([
    Dense(units=32,input_shape=(12,),activation='relu'),
    Dense(units=16,activation='relu'),
    Dense(units=2,activation='sigmoid')
])
```

```python
model.summary()
```

```python
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import BinaryCrossentropy
model.compile(optimizer=Adam(learning_rate=0.001),loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_train,y_train,batch_size=32,epochs=50,verbose=2)
```

```python
pred_y=model.predict(X_test)
```

```python
for i in pred_y:
  print(i)
```

```python
y_round=np.argmax(pred_y,axis=1)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
cm=confusion_matrix(y_test,y_round)
rep=classification_report(y_test,y_round)
```
